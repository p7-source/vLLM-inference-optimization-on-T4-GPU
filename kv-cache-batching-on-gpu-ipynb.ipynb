```python
MAX_SEQS = 4   # then 16, then 64

import time
from vllm import LLM, SamplingParams

prompts = [f"Write a short paragraph about topic number {i}." for i in range(64)]
params = SamplingParams(max_tokens=200, temperature=0)

llm = LLM(model="Qwen/Qwen2.5-3B-Instruct-AWQ", dtype="half",
          gpu_memory_utilization=0.8, max_model_len=2048,
          max_num_seqs=MAX_SEQS)
llm.generate(prompts[:2], params)
t = time.time()
outs = llm.generate(prompts, params)
n = sum(len(o.outputs[0].token_ids) for o in outs)
print(f"max_num_seqs={MAX_SEQS}: {n/(time.time()-t):.1f} tok/s")
```

```python
%%writefile dp_worker.py
import sys, time
from vllm import LLM, SamplingParams

part = int(sys.argv[1])
prompts = [f"Write a short paragraph about topic number {i}." for i in range(64)]
mine = prompts[part::2]                 # GPU 0 gets even, GPU 1 gets odd
params = SamplingParams(max_tokens=200, temperature=0)

llm = LLM(model="Qwen/Qwen2.5-3B-Instruct-AWQ", dtype="half",
          gpu_memory_utilization=0.8, max_model_len=2048, max_num_seqs=64)
llm.generate(mine[:2], params)          # warm-up
t = time.time()
outs = llm.generate(mine, params)
n = sum(len(o.outputs[0].token_ids) for o in outs)
print(f"GPU {part}: {n} tokens in {time.time()-t:.2f}s = {n/(time.time()-t):.1f} tok/s")
```
